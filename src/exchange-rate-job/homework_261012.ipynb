# 환율 API 데이터 조회
한국수출입은행 환율 API에서 GET 방식으로 환율 데이터를 가져옵니다.

```python
import requests
import json

url = "https://oapi.koreaexim.go.kr/site/program/financial/exchangeJSON"
auth_key = dbutils.secrets.get(scope = "grand_secret_scope", key = "AUTH_KEY")
print(auth_key)

"""
AP01 : 환율, AP02 : 대출금리, AP03 : 국제금리
authKey: "38fIqptdTHKN7NJVFg5spRgXTm47sFz2"
"""

searchdate = dbutils.widgets.get("SEARCHDATE")

params = {
    "authkey": auth_key,
    "searchdate": searchdate,
    "data": "AP01"
}

response = requests.get(url, params=params)

print(f"Status Code: {response.status_code}")
print(f"Response URL: {response.url}")



# data = response.json()
# print(json.dumps(data, indent=2, ensure_ascii=False))

data = response.json()
```

```python
from pyspark.sql.types import StructType, StructField, StringType, LongType, IntegerType

schema = StructType([
    StructField("result", LongType(), True),   # True → nullable 허용
    StructField("cur_unit", StringType(), True),
    StructField("ttb", StringType(), True),
    StructField("tts", StringType(), True),
    StructField("deal_bas_r", StringType(), True),
    StructField("bkpr", StringType(), True),
    StructField("yy_efee_r", StringType(), True),
    StructField("ten_dd_efee_r", StringType(), True),
    StructField("kftc_bkpr", StringType(), True),
    StructField("kftc_deal_bas_r", StringType(), True),
    StructField("cur_nm", StringType(), True)
])

df = spark.createDataFrame(data, schema=schema)
df.display()
```

```python
df.write.mode("overwrite").saveAsTable("workspace.default.exchange_rate")
```
